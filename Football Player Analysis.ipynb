#### Data cleaning:
#### - dropping nulls
#### - dropping duplicates 

```python
cl = pd.read_csv(r'C:\Users\Hatem\Desktop\players data md.csv')
cl.shape
```

```python
cl.dropna(inplace=True)
cl.shape
```

```python
cl.drop_duplicates(inplace=True)
cl.shape
```

```python
cl.nunique()
```

## Question 1:
### Correlation between weight, height, and age

#### correlation heatmap

```python
corr = cl[['weight_kg', 'height_cm','age']]
corr_matrix = corr.corr()
sns.heatmap(corr_matrix, annot=True)
plt.show()
```

## Data visualisation 

##  Question 2:
### What is the average age of football players?


### box plot

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x=cl['age'])
plt.title('Age Distribution')
plt.xlabel('Age')
plt.grid(True)
plt.show()
```

##  Question 3:
### How does age affect players wage?

### scatter plot

```python
plt.figure(figsize=(8, 8))
plt.scatter(cl['age'], cl['wage'])
plt.title('Age vs Wage')
plt.xlabel('Age')
plt.ylabel('Wage (EUR)')
plt.show()
```

## Question 4:
### Players height distribution 

### histogram

```python
plt.hist(cl.height_cm)
plt.title('height distribution')
plt.xlabel('height (cm)')
plt.ylabel('no. of players')
plt.show()
```

## Question 5:
### Preferred foot pizza chart :)

### pie chart

```python
#pie chart
left = cl.loc[cl['preferred_foot'] == 'Left'].count().iloc[0]
right = cl.loc[cl['preferred_foot'] == 'Right'].count().iloc[0]
labels = ['Left foot', 'Right foot']
colors = ['r', 'b']
plt.pie([left, right], labels=labels, colors=colors, autopct='%.f%%')
plt.title('Foot Preference of Players')
plt.show()
```

## Question 6:
### Most preferred positions for players

### countplot

```python
plt.figure(figsize=(16, 8))
sns.countplot(x='club_position', data=cl)
plt.title('position counts')
plt.show()
```
